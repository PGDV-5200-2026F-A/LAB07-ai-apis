# Python LLM API examples

```python
import json
import requests

from utils.keys import LLM_KEY
from utils.utils import build_prompt

# https://ai.google.dev/gemini-api/docs
GEMINI_URL = "https://generativelanguage.googleapis.com/v1beta/interactions"

GEMINI_HEADERS = {
  "Content-Type": "application/json",
  "x-goog-api-key": LLM_KEY,
}
```

```python
with open("../data/misc.json", "r") as ifp:
  misc_data = json.load(ifp)

with open("../data/astrolabe.json", "r") as ifp:
  astro_data = json.load(ifp)
```

## Miscellaneous

```python
misc_instructions = "If any of the items in the following list is a country, "\
  "output the name of the country and nothing else. If no item is a country, "\
  "output 'None'. If multiple items are countries, output the first one. Your "\
  "response is always one country name or 'None'. List: "
```

```python
row = misc_data[0]
locations = row["location"]

prompt = build_prompt(misc_instructions + str(locations))

res = requests.post(GEMINI_URL, json=prompt, headers=GEMINI_HEADERS)
data = res.json()
output = [o["content"] for o in data["steps"] if o["type"] == "model_output"][0]
print(output)
```

## Images

```python
astro_instructions = "Please describe the following image using nouns. "\
  "Use a plain-text list of just nouns. No escape characters. "\
  "No markdown, no html, no json. Just words and commas."
```

```python
astro_data[0]["first_image"]
```

```python
row = astro_data[0]
img_url = row["first_image"]

prompt = build_prompt(astro_instructions, img_url)
display(prompt)

res = requests.post(GEMINI_URL, headers=GEMINI_HEADERS, json=prompt)
data = res.json()
data["steps"][1]

output = [o["content"] for o in data["steps"] if o["type"] == "model_output"][0]
print(output[0]["text"])
```
